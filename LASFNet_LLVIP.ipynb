```python
import os
import shutil
import yaml

root_path = "/content/LASFNet/dataset/LLVIP"

def organize_split(modality, split):
    # Path folder saat ini (misal: /content/LASFNet/dataset/LLVIP/visible/train)
    current_dir = os.path.join(root_path, modality, split)
    if not os.path.exists(current_dir):
        print(f"Direktori {current_dir} tidak ditemukan.")
        return

    # Path target baru (misal: /content/LASFNet/dataset/LLVIP/visible/train/images)
    target_images_dir = os.path.join(current_dir, "images")

    # Jika folder 'images' belum ada, buat foldernya dan pindahkan gambar ke dalam
    if not os.path.exists(target_images_dir):
        os.makedirs(target_images_dir, exist_ok=True)
        print(f"Membuat folder target: {target_images_dir}")

        # Pindahkan file gambar saja (.jpg, .png)
        for item in os.listdir(current_dir):
            item_path = os.path.join(current_dir, item)
            if os.path.isfile(item_path) and item.lower().endswith(('.jpg', '.jpeg', '.png')):
                shutil.move(item_path, target_images_dir)
        print(f"Berhasil memindahkan gambar ke {target_images_dir}")
    else:
        print(f"Folder {target_images_dir} sudah siap!")

# Rapikan untuk visible dan infrared (train dan test)
for modality in ["visible", "infrared"]:
    for split in ["train", "test"]:
        organize_split(modality, split)

# Buat ulang file data/LLVIP.yaml dengan konfigurasi path yang tepat
yaml_content = {
    'path': '/content/LASFNet/dataset/LLVIP',
    'train_rgb': 'visible/train/images',
    'train_ir': 'infrared/train/images',
    'val_rgb': 'visible/test/images',
    'val_ir': 'infrared/test/images',
    'test_rgb': 'visible/test/images',
    'test_ir': 'infrared/test/images',
    'nc': 1,
    'names': ['person']
}

os.makedirs('/content/LASFNet/data', exist_ok=True)
with open('/content/LASFNet/data/LLVIP.yaml', 'w') as f:
    yaml.dump(yaml_content, f, sort_keys=False)

print("\nStruktur dataset berhasil disesuaikan dan LLVIP.yaml telah diperbarui!")
```

```python
import os
import xml.etree.ElementTree as ET

def convert_coor(size, box):
    dw = 1. / size[0]
    dh = 1. / size[1]
    x = (box[0] + box[1]) / 2.0
    y = (box[2] + box[3]) / 2.0
    w = box[1] - box[0]
    h = box[3] - box[2]
    return (x * dw, y * dh, w * dw, h * dh)

def convert_xml_to_yolo(xml_dir, images_dir, labels_dir):
    os.makedirs(labels_dir, exist_ok=True)
    if not os.path.exists(images_dir):
        print(f"Warning: {images_dir} tidak ditemukan.")
        return

    # Dapatkan semua nama file gambar tanpa ekstensi
    image_files = {os.path.splitext(f)[0] for f in os.listdir(images_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))}
    converted_count = 0

    for img_name in image_files:
        xml_path = os.path.join(xml_dir, f"{img_name}.xml")
        txt_path = os.path.join(labels_dir, f"{img_name}.txt")

        if not os.path.exists(xml_path):
            # Buat file label kosong jika tidak ada anotasi (background image)
            with open(txt_path, 'w') as f:
                pass
            continue

        tree = ET.parse(xml_path)
        root = tree.getroot()
        size = root.find('size')
        w = int(size.find('width').text)
        h = int(size.find('height').text)

        with open(txt_path, 'w') as out_file:
            for obj in root.iter('object'):
                cls = obj.find('name').text
                # Di LLVIP kelasnya adalah 'person' (ID: 0)
                cls_id = 0

                xmlbox = obj.find('bndbox')
                b = (float(xmlbox.find('xmin').text),
                     float(xmlbox.find('xmax').text),
                     float(xmlbox.find('ymin').text),
                     float(xmlbox.find('ymax').text))

                bb = convert_coor((w, h), b)
                out_file.write(f"{cls_id} " + " ".join([f"{a:.6f}" for a in bb]) + '\n')
        converted_count += 1

    print(f"Berhasil mengonversi & mencocokkan {converted_count} file label untuk {images_dir}")

# Eksekusi konversi label untuk tiap split
xml_source = "/content/LASFNet/dataset/LLVIP/Annotations"

# Karena modality visible dan infrared menggunakan gambar & koordinat yang sama,
# kita buat folder 'labels' di kedua modality agar dataloader LASFNet membacanya dengan benar
for modality in ["visible", "infrared"]:
    for split in ["train", "test"]:
        img_path = f"/content/LASFNet/dataset/LLVIP/{modality}/{split}/images"
        lbl_path = f"/content/LASFNet/dataset/LLVIP/{modality}/{split}/labels"
        convert_xml_to_yolo(xml_source, img_path, lbl_path)

print("\nSeluruh proses konversi anotasi selesai! Dataset sekarang siap digunakan.")
```

```python
import os
import shutil
import yaml

# Lokasi ekstraksi asli dan lokasi target
extracted_root = "/content/LASFNet/dataset/LLVIP"
nested_folder = os.path.join(extracted_root, "LLVIP")

print("Memeriksa struktur direktori...")

# Jika terdapat folder nested LLVIP di dalam LLVIP
if os.path.exists(nested_folder):
    print(f"Menemukan folder nested: {nested_folder}. Memindahkan isi folder ke {extracted_root}...")
    for item in os.listdir(nested_folder):
        source_path = os.path.join(nested_folder, item)
        target_path = os.path.join(extracted_root, item)
        if os.path.exists(target_path):
            if os.path.isdir(target_path):
                shutil.rmtree(target_path)
            else:
                os.remove(target_path)
        shutil.move(source_path, extracted_root)
    os.rmdir(nested_folder)
    print("Struktur folder berhasil dirapikan!")
else:
    print("Struktur folder sudah rapi atau tidak memerlukan perubahan.")

# Verifikasi keberadaan folder penting
paths_to_check = [
    "/content/LASFNet/dataset/LLVIP/visible/train/images",
    "/content/LASFNet/dataset/LLVIP/infrared/train/images"
]

for p in paths_to_check:
    if os.path.exists(p):
        print(f"Direktori ditemukan: {p}")
    else:
        print(f"Direktori TIDAK ditemukan: {p}")

# Update konfigurasi data/LLVIP.yaml kembali untuk memastikan keakuratan
yaml_content = {
    'path': '/content/LASFNet/dataset/LLVIP',
    'train_rgb': 'visible/train/images',
    'train_ir': 'infrared/train/images',
    'val_rgb': 'visible/test/images',
    'val_ir': 'infrared/test/images',
    'test_rgb': 'visible/test/images',
    'test_ir': 'infrared/test/images',
    'nc': 1,
    'names': ['person']
}

os.makedirs('/content/LASFNet/data', exist_ok=True)
with open('/content/LASFNet/data/LLVIP.yaml', 'w') as f:
    yaml.dump(yaml_content, f, sort_keys=False)

print("File data/LLVIP.yaml berhasil ditulis ulang dan diverifikasi!")
```

### Melanjutkan Training yang Terputus (Resume Training)

Jalankan cell di bawah ini untuk melanjutkan training menggunakan checkpoint terakhir (`last.pt`) yang tersimpan di Google Drive kamu.

```python
import os

# Patch kompatibilitas PyTorch 2.6+ untuk backward compatibility checkpoint YOLO
patch_script = """import torch
_old_load = torch.load
def _new_load(*args, **kwargs):
    kwargs.setdefault('weights_only', False)
    return _old_load(*args, **kwargs)
torch.load = _new_load
"""

files_to_patch = [
    "/content/LASFNet/train.py",
    "/content/LASFNet/val.py"
]

for file_path in files_to_patch:
    if os.path.exists(file_path):
        with open(file_path, "r") as f:
            content = f.read()

        if "_new_load" not in content:
            with open(file_path, "w") as f:
                f.write(patch_script + "\n" + content)
            print(f"✅ Berhasil me-patch: {file_path}")
        else:
            print(f"ℹ️ {file_path} sudah di-patch sebelumnya.")
    else:
        print(f"⚠️ File tidak ditemukan: {file_path}")
```
